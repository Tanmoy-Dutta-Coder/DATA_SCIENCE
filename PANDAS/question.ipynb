#### b)  Check sales_raw for missing values using .isnull().sum(). Fill missing quantity and unit_price with the column mean. Convert sale_date to a proper datetime column using to_datetime().   

```python
print(df1.isnull().sum())
print(df2.isnull().sum())
print(df3.isnull().sum())
```

```python
df1['quantity']=df1['quantity'].fillna(df1['quantity'].mean())
df1['unit_price']=df1['unit_price'].fillna(df1['unit_price'].mean())
print(df1[['quantity','unit_price']])

print(df1.isnull().sum())
```

```python
df1['Date_Time']=pd.to_datetime(df1['sale_date'],format='mixed')
print(df1['Date_Time'])
```

### c)  Merge sales_raw with stores on store_id using an inner join.
