```python
import pandas as pd
data={
    "Name":['Tanmoy','Ram','Kumar','Suras',
            'sam'],
    "Joining_Date":['2005-07-16','2005-08-31','2005-09-27','2005-09-12','2005-10-23'],
    # "Joining_Time":['14:23:51','02:12:55','23:45:12','12:22:05','01:45:20'],
    "Marks":[45,78,0,25,87]
    }
df=pd.DataFrame(data)
df["Joining_Date"]=pd.to_datetime(df['Joining_Date'])
df=df.set_index(df["Joining_Date"])
print(df)
```
